# Contacts

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import pandas as pd
Contacts = pd.read_excel('/content/drive/MyDrive/fin_project/not clean/Contacts.xlsx', dtype={'Id': str, 'Contact Owner Name': str})
print(Contacts.shape)
print()
summary_contacts = pd.DataFrame({
    'dtype': Contacts.dtypes,
    'non_null': Contacts.notna().sum(),
    'null': Contacts.isna().sum(),
    'null_%': (Contacts.isna().sum() / len(Contacts) * 100).round(1),
    'unique': Contacts.nunique()})
print(summary_contacts)
```

# ID

```python
# Id to string (for joins)
Contacts['Id'] = Contacts['Id'].astype(str)
print(Contacts['Id'].dtype)
print(Contacts['Id'].head(3).tolist())
```

# Created Time

```python
# Created Time to datetime
Contacts['Created Time'] = pd.to_datetime(Contacts['Created Time'], format='%d.%m.%Y %H:%M')
print(Contacts['Created Time'].dtype)
print(Contacts['Created Time'].head(3).tolist())
```

# Modified Time

```python
# Modified Time to datetime
Contacts['Modified Time'] = pd.to_datetime(Contacts['Modified Time'], format='%d.%m.%Y %H:%M')
print(Contacts['Modified Time'].dtype)
print(Contacts['Modified Time'].head(3).tolist())
```

# Saving

```python
Contacts.to_parquet('/content/drive/MyDrive/fin_project/Contacts_clean.parquet', index=False)
print("Contacts saved!")
```

# Descriptive Statistics

```python
# Check for duplicates in Contacts
print(f"Duplicates across all columns: {Contacts.duplicated().sum()}")
print(f"Duplicates by Id: {Contacts.duplicated(subset=['Id']).sum()}")
```

```python
# No numeric columns in Contacts - only categorical and date columns
print("General information:")
print(f"Total contacts: {len(Contacts)}")
print(f"Unique owners: {Contacts['Contact Owner Name'].nunique()}")
print()

# Distribution by owner
print("Top 10 Contact Owner Name:")
print(Contacts['Contact Owner Name'].value_counts().head(10))
```

```python
# Date analysis
print("Created Time:")
print(f"  First contact: {Contacts['Created Time'].min()}")
print(f"  Last contact: {Contacts['Created Time'].max()}")
print(f"  Period: {(Contacts['Created Time'].max() - Contacts['Created Time'].min()).days} days")
print()
print("Modified Time:")
print(f"  First modification: {Contacts['Modified Time'].min()}")
print(f"  Last modification: {Contacts['Modified Time'].max()}")
```

```python
# Monthly contact creation trend
Contacts['Created Month'] = Contacts['Created Time'].dt.to_period('M')
print("Contacts by month:")
print(Contacts['Created Month'].value_counts().sort_index())
```
